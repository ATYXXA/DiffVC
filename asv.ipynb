```python
import os
def search_wavs(rootdir):
    paths=[]
    for root, dirs, files in os.walk(rootdir):
        for file in files:
            if file.endswith(".wav"):
                paths.append(os.path.join(root,file))               
    
    return paths

def clear_cache(dir='audio_cache'):
    for i in os.listdir(dir):
        os.remove(f'{dir}/{i}')
```

```python
import os
def search_sub_wavs(dir):
    spks=os.listdir(dir)
    tgts=spks[1:]
    tgts.append(spks[0])
    paths=[]
    for i,spk in enumerate(spks):
        tgt=tgts[i]
        for utt in os.listdir(f'{dir}/{spk}'):
            if utt.endswith(".wav") and (tgt in utt) :
                paths.append(f'{dir}/{spk}/{utt}')  
    return paths

def get_oraclepath(path,oracledir='mos/oracle'):
    name=path.split('/')[-1]
    info=name.split('_')
    sourcespk=info[0]
    utt=info[-1][:3]
    return f'{oracledir}/{sourcespk}/{sourcespk}_{utt}.wav'

def search_sub_wavs1(dir):
    spks=os.listdir(dir)
    tgts=spks[1:]
    tgts.append(spks[0])
    paths=[]
    for i,spk in enumerate(spks):
        tgt=tgts[i]
        for utt in os.listdir(f'{dir}/{spk}'):
            if utt.endswith(".wav") and (tgt in utt) :
                paths.append(f'{dir}/{spk}/{utt}')  
                paths.append(get_oraclepath(f'{dir}/{spk}/{utt}'))
    return paths

def search_eg_wavs(rootdir):
    paths=[]
    for root, dirs, files in os.walk(rootdir):
        for file in files:
            if file.endswith(".wav"):
                paths.append(os.path.join(root,file))     
                paths.append(get_oraclepath(os.path.join(root,file)))          
    
    return paths
```

```python
from tqdm import tqdm
def batch_asv(dir,threshold=0.25):
    clear_cache()
    preds=[]
    scores=[]
    for wav in tqdm(search_wavs(dir)):
        oracle=get_oraclepath(wav)
        # oracle=get_oraclepath1(wav)
        score, prediction = verification.verify_files(wav, oracle, threshold=threshold)
        score = score.detach().cpu().item()
        prediction = prediction.detach().cpu().item()
        preds.append(int(prediction))
        scores.append(score)
    return 1 - sum(preds)/len(preds)
```

```python
def create_veri_test(data_folder,mode='all'):
    veri_test = []
    if mode=='all':
        dev_save_path = search_wavs(data_folder)
    elif mode=='sub':
        dev_save_path = search_sub_wavs(data_folder)
    elif mode=='rand':
        dev_save_path = search_rand_wavs(data_folder)
    elif mode=='sub1':
        dev_save_path = search_sub_wavs1(data_folder)
    elif mode=='eg':
        dev_save_path = search_eg_wavs(data_folder)

    for i in range(0, len(dev_save_path) - 1):
        index = i + 1
        while index <= len(dev_save_path) - 1:
            if dev_save_path[i].split("/")[-2] != dev_save_path[index].split("/")[-2]:
                veri_test.append("{} {} {}".format(0, dev_save_path[i], dev_save_path[index]))
            else:
                veri_test.append("{} {} {}".format(1, dev_save_path[i], dev_save_path[index]))
            index+=1
    return veri_test
create_veri_test('mos/oracle',mode='rand')
```

```python
import torch
from tqdm import tqdm
def get_verification_scores(dir,mode='all'):
    """Computes positive and negative scores given the verification split."""
    scores = []
    positive_scores = []
    negative_scores = []

    veri_test = create_veri_test(dir,mode)

    for  line in tqdm(veri_test):
        # Reading verification file (enrol_file test_file label)
        lab_pair = int(line.split(" ")[0])
        enrol_id = line.split(" ")[1]
        test_id = line.split(" ")[2]

        # Compute the score for the given sentence
        # score = similarity(enrol, test)[0]
        score, pred = verification.verify_files(enrol_id,test_id)
        score = score.detach().cpu().item()
        # pred = pred.detach().cpu().item()        

        scores.append(score)

        if lab_pair == 1:
            positive_scores.append(score)
        else:
            negative_scores.append(score)

    return positive_scores, negative_scores
```

```python
import librosa
import soundfile as sf
import os

def rs16k(input_path, cache_dir = 'audio_cache'):
    audio_data, original_sr = librosa.load(input_path, sr=None)
    resampled_audio = librosa.resample(audio_data, orig_sr=original_sr, target_sr=16000)
    output_path = f'{cache_dir}/{os.path.basename(input_path)}'
    sf.write(output_path, resampled_audio, 16000)
    return output_path

def get_score_ms(cvtpath_, tgtpath_):
    cvtpath = rs16k(cvtpath_)
    tgtpath = rs16k(tgtpath_)
    result = sv_pipeline([cvtpath, tgtpath])
    os.remove(cvtpath)
    os.remove(tgtpath)
    return result['score']
```
