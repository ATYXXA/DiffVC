```python
import torch
import librosa
from tqdm import tqdm
def get_mos(wavpath):
    wave, sr = librosa.load(wavpath, sr=None, mono=True)
    score = mospredictor(torch.from_numpy(wave).unsqueeze(0), sr)
    return score.item()
```

```python
import os
def search_wavs(rootdir):
    paths=[]
    for root, dirs, files in os.walk(rootdir):
        for file in files:
            if file.endswith(".wav") or file.endswith(".mp3"):
                paths.append(os.path.join(root,file))
    
    return paths
```

```python
import numpy as np
from tqdm import tqdm
def batch_mos(dir):
    allmos=[]
    for wav in tqdm(search_wavs(dir)):
        mos = get_mos(wav)
        allmos.append(mos.item())
    return np.mean(allmos), allmos
```

```python
import librosa
import os
import soundfile as sf


def rs_dir(input_folder, output_folder):

    for root, dirs, files in os.walk(input_folder):
        # 构建输出文件夹对应的子目录路径
        relative_path = os.path.relpath(root, input_folder)
        output_subfolder = os.path.join(output_folder, relative_path)
        # 如果输出子目录不存在，则创建它
        if not os.path.exists(output_subfolder):
            os.makedirs(output_subfolder)

        # 遍历当前子目录下的所有文件
        for file_name in files:
            file_path = os.path.join(root, file_name)
            try:
                # 使用librosa加载音频文件，获取音频数据和原始采样率
                audio_data, sr = librosa.load(file_path, sr=None)
                # 将音频数据重采样到16000Hz（16kHz）
                audio_data_resampled = librosa.resample(audio_data, orig_sr=sr, target_sr=16000)
                # 构建输出文件的路径
                output_file_path = os.path.join(output_subfolder, file_name)
                # 使用soundfile库将重采样后的音频数据保存为新的音频文件
                sf.write(output_file_path, audio_data_resampled, 16000)
            except Exception as e:
                print(f"处理文件 {file_name} 时出现错误: {e}")
```
