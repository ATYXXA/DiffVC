```python
win_len = 10 # seconds
mosmodel = speechmetrics.load('absolute.mosnet', win_len)
def get_mos(path):
    score = mosmodel(path)['mosnet'].item()
    return score
get_mos('mos/oracle/p238/p238_001.wav')
```

```python
import os
def search_wavs(rootdir):
    paths=[]
    for root, dirs, files in os.walk(rootdir):
        for file in files:
            if file.endswith(".wav"):
                paths.append(os.path.join(root,file))
    
    return paths
```

```python
import numpy as np
from tqdm import tqdm
def batch_mos(dir):
    allmos=[]
    for wav in tqdm(search_wavs(dir)):
        mos = get_mos(wav)
        allmos.append(mos)
    return np.mean(allmos), allmos
```

```python
mos, _ = batch_mos('mos/oracle/')
mos
```

```python
mosmap={}

for dir in ['mos/oracle/',
            'mos/outputs_diffvc/',
            'mos/outputs_diffvcplus/',
            'mos/outputs_diffvcp_light/',
            'mos/outputs_diffvcp_ctrl/']:
    mos, _ = batch_mos(dir)
    mosmap[dir]=mos
    print(mos)
mosmap
```
